# 🐼 Pandas `query()` – Quick Notes

- `query()` filters **DataFrame rows** using a **string condition written as a string expression**.   

---

## 🔹 Core Rules

- Condition must be a **string**: `"age > 25"`. 
- Use **column names directly** (no `df['col']` inside the expression).   
- Use **backticks** for columns with spaces or special characters, e.g. `` `total marks` ``.   
- Supported operators: `&` (and), `|` (or), `~` (not), `>`, `>=`, `<`, `<=`, `==`.
- Use **`@`** to reference Python variables in the expression. 
- **Chained comparisons** like `50 <= score <= 90` are allowed.

---

## ✅ Advantages

- Cleaner, more readable filtering syntax. 
- Avoids repetitive `df[...]` patterns.  
- Works well with complex, multi-condition filters. 


```python
#filtering data using query

new2=data.query("aqi == 100 & temperature >30 | country == 'US'")[["city","country","aqi"]]

aqi_value=60
new3=data.query("aqi == @aqi_value & temperature >30 & country == 'US'")[["city","country","aqi"]]  #using variable with @
print(new3)
```
