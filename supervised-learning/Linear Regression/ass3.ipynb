```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
data=pd.read_csv("Iris.csv")

data['Species'].unique()
```

```python
data['Species']=pd.factorize(data['Species'])[0]
```

```python
data
```

```python
X = data.drop("Species", axis=1)
y = data["Species"]

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.5, random_state=42
)
X_test=X
y_test=y
```

```python
knn=LogisticRegression(max_iter=1000)
knn.fit(X_train,y_train)
knn.predict(X_test)
```

```python
lr=KNeighborsClassifier(n_neighbors=7)
lr.fit(X_train,y_train)
lr.predict(X_test)
```

```python
gnb = GaussianNB()
gnb.fit(X_train, y_train)
gnb.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

models = {'Logistic Regression': lr, 'KNN': knn, 'Naive Bayes': gnb}

for name, model in models.items():
    y_pred = model.predict(X)
    print(f"\n{name}")
    print("Accuracy:", accuracy_score(y, y_pred))
    
    print("cm:", confusion_matrix(y, y_pred))

    print("classification report:", classification_report(y, y_pred))
```
