```python
from sklearn.datasets import make_blobs
```

```python
X,y=make_blobs(
    n_samples=1000, 
    centers=4,
    n_features=2,
    random_state=42
)
```

```python
import seaborn as sns
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1])
```

```python
from sklearn.cluster import KMeans
```

```python
model=KMeans(
    n_clusters=4,
    random_state=42
)
```

```python
label=model.fit_predict(X) #desides cluster for each data points
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],c=label)
```

### find K value

```python
#elbow method(manual)

wcss=[]

for k in range(1,21):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)


sns.lineplot(x=range(1,21),y=wcss,markers=True,marker="o")
```

```python
#elbow method(using library)
from kneed import KneeLocator

kl = KneeLocator(range(1,21), wcss, curve="convex", direction="decreasing")

print(kl.knee)
```

### Silhouette Score

```python
from sklearn.metrics import silhouette_score
```

```python
ss=[]

for k in range(2,21):  #k range must not be starting from 1
    kmeans1=KMeans(n_clusters=k,random_state=42)
    labels=kmeans1.fit_predict(X)
    score=silhouette_score(X,labels)
    ss.append(score)
sns.lineplot(x=range(2,21),y=ss,markers=True,marker="o")  #highest = better K value
```
